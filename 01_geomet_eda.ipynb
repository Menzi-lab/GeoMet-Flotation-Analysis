```python

from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Correct path to your data folder
DATA_DIR = Path(r"C:\Users\Menzi\Desktop\GeoMet-Flotation-Analysis\Data")

# Load the three datasets
flotation = pd.read_csv(DATA_DIR / "flotation.csv")
comminution = pd.read_csv(DATA_DIR / "comminution.csv")
drillholes = pd.read_csv(DATA_DIR / "drillholes.csv")

print("All datasets loaded successfully!")
print("Flotation:", flotation.shape)
print("Comminution:", comminution.shape)
print("Drill holes:", drillholes.shape)

```

```python

datasets = {
    "Flotation": flotation,
    "Comminution": comminution,
    "Drill holes": drillholes
}

for name, df in datasets.items():
    print("\n" + "=" * 70)
    print(f"{name.upper()} DATASET")
    print("=" * 70)

    print(f"Rows: {df.shape[0]:,}")
    print(f"Columns: {df.shape[1]}")

    print("\nColumn names and data types:")
    display(df.dtypes.rename("Data type").to_frame())

    print("\nFirst five rows:")
    display(df.head())

    print("\nMissing values per column:")
    display(df.isna().sum().to_frame("Missing values"))
```

Checking whether the numbers from these csv files can be linked or not


```python

# Check identifiers and duplicate records
for name, df in datasets.items():
    print(f"\n{'=' * 50}")
    print(name.upper())

    print("Total rows:", len(df))
    print("Unique HOLEID values:", df["HOLEID"].nunique())
    print("Missing HOLEID values:", df["HOLEID"].isna().sum())
    print("Duplicate complete rows:", df.duplicated().sum())
    print("Repeated HOLEID rows:", df["HOLEID"].duplicated().sum())
```

```python

flotation_ids = set(flotation["HOLEID"].dropna())
comminution_ids = set(comminution["HOLEID"].dropna())
drillhole_ids = set(drillholes["HOLEID"].dropna())

print("Flotation IDs also in drill holes:",
      len(flotation_ids & drillhole_ids))

print("Comminution IDs also in drill holes:",
      len(comminution_ids & drillhole_ids))

print("Flotation IDs also in comminution:",
      len(flotation_ids & comminution_ids))
```

```python

# Compare coordinates for HOLEIDs shared between datasets

pairs = [
    ("Flotation", flotation, "Drill holes", drillholes),
    ("Comminution", comminution, "Drill holes", drillholes),
    ("Flotation", flotation, "Comminution", comminution),
]

for name_a, df_a, name_b, df_b in pairs:
    print(f"\n{'=' * 60}")
    print(f"{name_a} vs {name_b}")

    shared_ids = set(df_a["HOLEID"]) & set(df_b["HOLEID"])

    print("Shared HOLEIDs:", len(shared_ids))

    # Show coordinate records for a few shared holes
    sample_ids = sorted(shared_ids)[:5]

    for hole_id in sample_ids:
        print(f"\nHOLEID: {hole_id}")

        print(f"{name_a}:")
        display(
            df_a.loc[
                df_a["HOLEID"] == hole_id,
                ["HOLEID", "X", "Y", "Z"]
            ].head(5)
        )

        print(f"{name_b}:")
        display(
            df_b.loc[
                df_b["HOLEID"] == hole_id,
                ["HOLEID", "X", "Y", "Z"]
            ].head(5)
        )
```
